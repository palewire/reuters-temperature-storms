```python
import pandas as pd
```

```python
ranking_url = "https://graphics.thomsonreuters.com/storm-outputs/rts/newsapps_reuters-climate-monitor/heat-storms/by-lifetime-area.json"
```

```python
live = pd.read_json(ranking_url).rename(columns={"startDate": "start_date"})
```

```python
live["rank"] = range(1, len(live) + 1)
```

```python
catalog = pd.read_parquet("data/all_storms.parquet")
```

```python
catalog["start_date"] = catalog["start_date"].dt.strftime("%Y-%m-%d")
```

```python
largest = live.merge(catalog, on=["name", "start_date"], how="left", validate="one_to_one")
```
